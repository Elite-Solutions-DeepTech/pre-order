##  Question 1: Anagram Grouper


##  Question 1: Anagram Grouper

###  My Explanation:

In this task, I created a function that groups together words that are anagrams.

What is an anagram?  
An anagram is a word made by rearranging the letters of another word.  
For example:  
- "eat", "tea", and "ate" are anagrams because they all use the same letters.

---

How I Thought About It:

- First, I realized that sorting the letters of each word would help me find which words are anagrams.
  For example, sorting "eat" gives "aet"
- Sorting "tea" also gives "aet"
- So if two words give the same sorted result, they must be anagrams.

---

How the Code Works:

 1. I used a dictionary to group the words.
 2. For each word in the list:
 - I sorted the word to get a common key.
 - I added the word to the dictionary using that key.
 3. At the end, I returned just the values of the dictionary, which are the anagram groups.

---

 ## What the User Can Do:

The user can type any number of words (separated by spaces), and the function will group the anagrams automatically.

For example, if the
input is:  eat tea tan ate nat bat
output:[['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]

This is how I solved and understood the Anagram Grouper problem.
          


```python
def group_anagrams(words): # Empty dictionary to hold groups
  anagram_dict={}
  for word in words:
    key=''.join(sorted(word)) # creating a key using sorted letters
    if key in anagram_dict:
      anagram_dict[key].append(word) # adding to existing list
    else:
      anagram_dict[key]=[word] # Start a  new list
    return list(anagram_dict.values()) # returns grouped lists
input_string=input("Enter words separated by spaces:") #Dynamic input from user
words_list=input_string.split() #Converting into lists
result=group_anagrams(words_list) # Calling the function
print("Grouped Anagrams:",result) # Displaying Output
```

##Question 2 :Password Srength  Checker






```python


def is_strong_password(password):
    # Check if the password is at least 8 characters long
    if len(password) < 8:
        return False
    # Initialize flags
    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False
    # Define special characters manually
    special_characters = "!@#$%^&*()-_=+[]{}|;:',.<>?/`~"
    # Check each character in the password
    for char in password:
        if 'A' <= char <= 'Z':
            has_upper = True
        elif 'a' <= char <= 'z':
            has_lower = True
        elif '0' <= char <= '9':
            has_digit = True
        elif char in special_characters:
            has_special = True
    # Return True only if all conditions are met
    return has_upper and has_lower and has_digit and has_special
# Input from user
password = input("Enter a password to check: ")
# Output result
if is_strong_password(password):
    print(" Strong password!")
else:
    print(" Weak password. Please ensure it has:")
    print("- At least 8 characters")
    print("- At least 1 uppercase letter")
    print("- At least 1 lowercase letter")
    print("- At least 1 digit")
    print("- At least 1 special character (e.g., !@#$%)")
```

##  Question 3: Transaction Log Analyzer


```python



def transaction_log_analyzer(transactions):
    result = {}
    for txn in transactions:
        if txn['type'] == 'debit':
            user = txn['user']
            amount = txn['amount']
            if user in result:
                result[user] += amount
            else:
                result[user] = amount
    return result

# Taking input from the user
transactions = []
n = int(input("Enter number of transactions: "))
for i in range(n):
    print(f"\nEnter details for transaction {i + 1}:")
    user = input("User name: ")
    amount = float(input("Amount: "))
    txn_type = input("Type (debit/credit): ").strip().lower()
    # Append the transaction as a dictionary
    transactions.append({
        'user': user,
        'amount': amount,
        'type': txn_type
    })

# Analyzing the transactions
output = transaction_log_analyzer(transactions)

# Display the result
print("\nTotal debit amount spent by each user:")
for user, total in output.items():
    print(f"{user}: ₹{total}")
```

##  Question 4: Matrix Rotation


```python

def rotate_matrix(matrix):
    n = len(matrix)
    # Step 1: Transpose the matrix
    for i in range(n):
        for j in range(i + 1, n):
            matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
    # Step 2: Reverse each row
    for row in matrix:
        row.reverse()
    return matrix
# Get input from the user
n = int(input("Enter the size of the matrix (N x N): "))
print(f"Enter {n} rows with {n} elements each (separated by spaces):")
matrix = []
for i in range(n):
    row = list(map(int, input(f"Row {i + 1}: ").split()))
    if len(row) != n:
        print("Error: Please enter exactly", n, "numbers.")
        exit()
    matrix.append(row)
print("\nOriginal Matrix:")
for row in matrix:
    print(row)
# Rotate the matrix
rotated = rotate_matrix(matrix)
print("\nRotated Matrix (90 degrees clockwise):")
for row in rotated:
    print(row)

```

##  Question 5: Circular Dependency Detector


```python


def detect_circular_dependency(dependencies):
    from collections import defaultdict
    # Create adjacency list
    graph = defaultdict(list)
    for src, dst in dependencies:
        graph[src].append(dst)
    visited = set()
    rec_stack = set()
    def dfs(node, path):
        visited.add(node)
        rec_stack.add(node)
        path.append(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                if dfs(neighbor, path):
                    return True
            elif neighbor in rec_stack:
                # Cycle detected, extract the cycle
                cycle_start_index = path.index(neighbor)
                cycle = path[cycle_start_index:]
                result.extend(cycle)
                return True
        path.pop()
        rec_stack.remove(node)
        return False
    result = []
    for node in graph:
        if node not in visited:
            if dfs(node, []):
                break
    return result
```

##  Question 6: Schedule Optimizer


```python


def schedule_optimizer(meetings):
    # Step 1: Sort meetings by their end times
    meetings.sort(key=lambda x: x[1])
    # Step 2: Initialize counters
    count = 0
    last_end_time = 0
    # Step 3: Loop through meetings
    for start, end in meetings:
        if start >= last_end_time:
            count += 1
            last_end_time = end
    return count
# 🔽 Taking user input
n = int(input("Enter the number of meetings: "))
meetings = []
print("Enter each meeting's start and end time (space-separated):")
for i in range(n):
    s, e = map(int, input(f"Meeting {i+1}: ").split())
    meetings.append((s, e))
# ✅ Output the result
max_meetings = schedule_optimizer(meetings)
print("Maximum number of non-overlapping meetings:", max_meetings)
```

###  Schedule Optimizer

 Problem Statement:  
We are given a list of meetings with start and end times. Our goal is to find out the **maximum number of non-overlapping meetings** we can attend.

---

###  My Approach

 Here's how I solved this problem in a simple and smart way:

1. I thought of each meeting as a time slot with a start and end.
2. To make sure I can attend the **maximum number** of meetings, I decided to always pick the one that **ends the earliest** — so that I have more time left to attend future meetings.
3. So, I **sorted the meetings by their end time**.
4. Then, I used a variable called `last_end_time` to track the ending time of the last meeting I attended.
5. I looped through each meeting:
- If the current meeting’s start time is **after or equal to** the `last_end_time`, it means it doesn't overlap — so I updated `last_end_time`.
6. Finally, I printed how many such meetings I could attend.
